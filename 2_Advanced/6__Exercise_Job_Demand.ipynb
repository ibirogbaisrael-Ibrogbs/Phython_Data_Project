```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_us = df[df['job_country'] == 'United States'].copy()
```

```python
df_us
```

```python
df_us['job_posted_month'] = df_us['job_posted_date'].dt.strftime('%B')
```

```python
df_us.head(2)
```

```python
df_us_pivot = df_us.pivot_table(index='job_posted_month', columns='job_title_short', aggfunc='size')
```

```python
df_us_pivot.reset_index(inplace=True)
df_us_pivot['month_no'] = pd.to_datatime
```

```python
import pandas as pd

# define correct month order
month_order = [
    'January', 'February', 'March', 'April', 'May', 'June',
    'July', 'August', 'September', 'October', 'November', 'December'
]

# convert to ordered categorical
df_us['job_posted_month'] = pd.Categorical(
    df_us['job_posted_month'],
    categories=month_order,
    ordered=True
)

# now create pivot table
df_us_pivot = df_us.pivot_table(
    index='job_posted_month',
    columns='job_title_short',
    aggfunc='size'
)

# ensure proper sorting
df_us_pivot = df_us_pivot.sort_index()
```

```python
df_us_pivot
```

```python
df_us_pivot.plot(kind='line')
```

```python
top_3 = df_us['job_title_short'].value_counts().head(3)
top_3 = top_3.index.tolist()
```

```python
df_us_pivot[top_3].plot()
plt.ylabel('Job Count')
plt.xlabel('2023')
plt.title('Monthly Job Postings for Top Data Jobs in the Us')

```
