```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
from datasets import load_dataset
import ast


dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df['job_work_from_home'].value_counts().plot(kind='pie', startangle=180)
plt.title('Work from Home Status')

plt.show()
```

```python
df['job_title_short'].value_counts().plot(kind='pie', autopct='%1.1f%%')
```

```python

fig, ax = plt.subplots(1, 3)

dict_column = {
    'job_work_from_home': 'Work from Home',
    'job_no_degree_mention': 'Job Degree Reg.',
    'job_health_insurance': 'Health Insurance Offered' 
}

for i, (column, title) in enumerate(dict_column.items()):
    counts = df[column].value_counts()
    
    ax[i].pie(counts, labels=counts.index, startangle=90, autopct='%1.1f%%')
    ax[i].set_title(title)

plt.show()
```
